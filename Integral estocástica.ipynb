```python
import numpy as np
import random as r
import matplotlib.pyplot as plt
```

```python
a=0
b=1
n=1000
```

```python
def brown(T,L):
  lista=[]
  w0=0
  dt=T/L
  for i in range (1,L):
    dw=np.random.normal(0,1)
    w_1=w0+np.sqrt(dt)*dw
    lista.append(w_1)
    w0=w_1
  lista.insert(0,0)
  return lista
```

```python
def integral(a,b,n):
    inte=0
    t=np.linspace(a,b,n)
    t=t.tolist()
    B=brown(b,n)
    for i in range (1,n):
      inte+= B[i-1]*(t[i]-t[i-1])
    return inte
```

```python
integral(0,1,100)
```

```python
plt.plot(B, color="orange")
```

```python
valores=[]
for i in range(1,1000):
  avr=integral(0,1,100)
  m=avr/i
  valores.append(m)
```

```python
np.mean(valores)
```

```python
np.var(valores)
```
