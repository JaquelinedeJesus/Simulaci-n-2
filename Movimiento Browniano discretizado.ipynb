# **Movimiento Browniano discretizado**

```python
import random as r
import numpy as np
import matplotlib.pyplot as plt
```

```python
def brown(T,n):
  lista=[]
  dt=np.sqrt(T/n)
  w_0=0
  for i in range(1,n):
    dw = r.gauss(0,1)
    w_1=w_0+dt*dw
    lista.append(w_1)
    w_0=w_1
  return lista
```

```python
M = brown(1,1000)
```

```python
plt.plot(M)
```
